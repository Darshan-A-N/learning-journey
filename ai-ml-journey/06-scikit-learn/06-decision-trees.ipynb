#  Decision Trees

## The Idea
Like a flowchart of questions:
- Is age &gt; 30? → Yes/No
- Is salary &gt; 50k? → Yes/No
- → Prediction: Buy or Don't Buy

## Key Terms
- **Root Node**: First question
- **Leaf Node**: Final prediction
- **Split**: Dividing data based on a condition
- **Entropy**: Measure of chaos (0 = pure, 1 = chaotic)

```python
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.model_selection import train_test_split
import matplotlib.pyplot as plt
import pandas as pd

# Create sample data: Play tennis based on weather
data = {
    "Outlook": ["Sunny", "Sunny", "Overcast", "Rain", "Rain", "Rain", 
                "Overcast", "Sunny", "Sunny", "Rain", "Sunny", 
                "Overcast", "Overcast", "Rain"],
    "Humidity": ["High", "High", "High", "High", "Normal", "Normal",
                 "Normal", "High", "Normal", "Normal", "Normal",
                 "High", "Normal", "High"],
    "Wind": ["Weak", "Strong", "Weak", "Weak", "Weak", "Strong",
             "Strong", "Weak", "Weak", "Weak", "Strong",
             "Strong", "Weak", "Strong"],
    "Play": ["No", "No", "Yes", "Yes", "Yes", "No",
             "Yes", "No", "Yes", "Yes", "Yes",
             "Yes", "Yes", "No"]
}

df = pd.DataFrame(data)
print(df)

# Convert to numbers (Decision Trees need numbers)
df_encoded = pd.get_dummies(df.drop("Play", axis=1))
y = df["Play"].map({"Yes": 1, "No": 0})

print("\nEncoded features:\n", df_encoded.head())
print("\nTarget:\n", y.head())
```

# Train Decision Tree
tree = DecisionTreeClassifier(max_depth=3, random_state=42)
tree.fit(df_encoded, y)

# Visualize the tree
plt.figure(figsize=(20, 10))
plot_tree(tree, feature_names=df_encoded.columns, 
          class_names=["No", "Yes"], filled=True, rounded=True)
plt.title("Decision Tree: Play Tennis?")
plt.show()

```python
# Predict: Should we play today?
# Today: Sunny, Humidity=Normal, Wind=Weak
today = pd.DataFrame({
    "Outlook_Overcast": [0],
    "Outlook_Rain": [0],
    "Outlook_Sunny": [1],
    "Humidity_High": [0],
    "Humidity_Normal": [1],
    "Wind_Strong": [0],
    "Wind_Weak": [1]
})

prediction = tree.predict(today)
probability = tree.predict_proba(today)

print(f"Play Tennis? {'Yes' if prediction[0] == 1 else 'No'}")
print(f"Confidence: {probability[0][prediction[0]]:.2%}")
```
