```python
def greet(name):
    return f"Hello, {name}!"

def calculate_bmi(weight, height):
    return weight / (height ** 2)

# Test them
print(greet("Alice"))
print(f"BMI: {calculate_bmi(70, 1.75):.2f}")
```

```python
student = {
    "name": "Alice",
    "branch": "ECE",
    "year": 2,
    "skills": ["Python", "HTML", "CSS"]
}

# Access, add, update
print(student["name"])
student["cgpa"] = 8.5
student["skills"].append("NumPy")

# Loop through
for key, value in student.items():
    print(f"{key}: {value}")
```

```python
# Create a text file
with open("my_data.txt", "w") as file:
    file.write("Name: Alice\n")
    file.write("Branch: ECE\n")

# Read it back
with open("my_data.txt", "r") as file:
    content = file.read()
    print(content)
```

```python
# Create a function that takes a list of students (dicts)
# and saves them to a file
def save_students(students, filename):
    with open(filename, "w") as file:
        for student in students:
            file.write(f"{student['name']},{student['branch']},{student['cgpa']}\n")

students = [
    {"name": "Alice", "branch": "ECE", "cgpa": 8.5},
    {"name": "Bob", "branch": "CSE", "cgpa": 9.0}
]

save_students(students, "students.csv")
print("Saved!")
```
